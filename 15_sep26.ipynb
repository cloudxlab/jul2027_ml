```python
def diff(f, x):
    eps = 1e-10
    d = x * 1e-10
    if d == 0:
        d = eps
    # return (f(x + d) - f(x))/d
    # return (f(x) - f(x-d))/d
    return (f(x+d) - f(x-d))/(2*d)
```
