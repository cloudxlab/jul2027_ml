```python
class Node():
    def __init__(self, num):
        self.num = num 
        self.left = None 
        self.right = None
    def insert(self, new_node):
        if new_node.num <= self.num:
            if self.left:
                self.left.insert(new_node)
            else:
                self.left = new_node
        else:
            if self.right:
                self.right.insert(new_node)
            else:
                self.right = new_node
    def inorder(self):
        result = []
        if self.left:
            result.extend(self.left.inorder())
        result.append(self.num)
        if self.right:
            result.extend(self.right.inorder())
        return result
    def decreasing(self):
        result = []
        if self.right:
            result.extend(self.right.decreasing())
        result.append(self.num)
        if self.left:
            result.extend(self.left.decreasing())    
        return result
    
```

```python
def create_insert(root, num):
    root.insert(Node(num))
```

```python
root = Node(1)
create_insert(root, -1)
create_insert(root, 5)
create_insert(root, -5)
create_insert(root, 4)
```
