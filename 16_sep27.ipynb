```python
# standard deviation?
dataset1 = [1, 1.1, .9]
dataset2 = [1, -5, 10] 

# Which dataset is has more variance? 
# Come up with a number representing variance

# subtract each number from mean, take the square, sum it up and divide by counts. -> variance

# take the variance -> square root -> standard deviation

np.std(dataset1), np.std(dataset2)
```

```python
# Can you do it in 2D?

def compute_mean(numbers):
    sm = 0
    counts = 0
    for i in numbers:
        sm += i 
        counts += 1
    return sm / counts

from math import sqrt
def compute_sd(numbers):
    mu = compute_mean(numbers)
    sd = 0
    var = 0
    n = len(numbers)
    for i in numbers:
        var += (i-mu)**2
        #sd = sd+compute_mean(abs(sd1))
    sd = sqrt(var/n)
    return sd
#compute_sd([0, 100, 0, 100, 50])
compute_sd([10, 12, 12, 13, 12, 11, 90])
```

```python
compute_sd(dataset1), compute_sd(dataset2)
```
