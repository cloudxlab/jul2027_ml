```python
def solve_for_first_variable(equation,vars):
    val = []
    val2 = []
    final_s =[]
    #We can use slicing as well
    #middle = equation[1:-1] : beech ke saare coefficients (x2, x3... ke)
    for i,j in enumerate(equation):
        if i in range(1,(len(equation))-1):
            val.append(j)
        else:
            val2.append(j)
    #print(val2)
    for x,y in zip(val,vars):
        
        #print("hi")
        if y != []:
            #print(j)
            s = x*y
            final_s.append(s)
    m,n = val2
    #print(final_s)
    
    d = 0
    for r in final_s:
        d = d+r
    result = (n-d)/m
        
    return result
solve_for_first_variable([3, 4, 6, 20], [5, 6])
```
