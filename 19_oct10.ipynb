```python
def find_min(arr):
    mn = arr[0]
    for e in arr:
        if e < mn:
            mn = e
    return mn 

find_min([1,2,-3, 5])
```

```python
def print_n(n):
	total = 0
	for i in range(n):
		for j in range(n):
			print(i, j, 'hello')
			total += 1
	print("total: ", total)
```
