```python
# Define a class that takes num and left and right in the __init__ method. 
class Node():
    def __init__(self, num, left=None, right=None):
        self.num = num 
        self.left = left
        self.right = right
    def __repr__(self):
        return str(self.num)
    def sum(self):
        total = self.num 
        if self.left:
            total += self.left.sum()
        if self.right:
            total += self.right.sum()
        return total 
```
