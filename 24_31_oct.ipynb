```python
txt = '''A person is what A person does. I am a person.
I like to code.
'''
print(txt)
words = txt.split()
```

```python
words 
```

```python
d = {} # word:freq...
for word in words:
    print(f"Processing word: {word}")
    if word in d:
        print("Word exists.. increasing it")
        d[word] = d[word] + 1
    else:
        print("Word doesn't exist save the freq as 1")
        d[word] = 1
    print(f"Dict: {d}")
```

```python
# Exercise

# 1. find the first factor 

# 48 -> 2
# 49 -> 7
# 169 -> 13
# 
def first_factor(num):
    for f in range(2, num):
        if num % f == 0:
            return f
    return num
        
```
