```python
# 1 from A to C using B as aux

def move2(src, target, aux):
    print(f"Moving 1 from {src} to {aux}")
    print(f"Moving 2 from {src} to {target}")
    print(f"Moving 1 from {aux} to {target}")
```

```python
def move(n, src, target, aux):
    if n==0:
        return
    move(n-1, src, aux, target)
    print(f"Moving {n} from {src} to {target}")
    move(n-1, aux, target, src)
    
```

```python
move(3, 'A', 'C', 'B')
```

```python
def move(n, src, target, aux, indent_str=''):
    print(f"{indent_str}Solving {n} src: {src}, target: {target}, aux: {aux}")
    if n==0:
        return
    move(n-1, src, aux, target, indent_str + "\t")
    print(f"{indent_str}Moving {n} from {src} to {target}")
    move(n-1, aux, target, src, indent_str + "\t")
    
```

```python
move(3, 'A', 'C', 'B')
```
